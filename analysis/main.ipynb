```python
import pandas as pd
import sqlite3

# Read sqlite query results into a pandas DataFrame
con = sqlite3.connect("../data/db.db")
df = pd.read_sql_query("SELECT * from CarPostings", con)

# Verify that result of SQL query is stored in the dataframe
print(df.head())

con.close()
```

```python
# Analysis by brand
import plotly.express as px

df1 = df[["car_brand", "price", "mileage", "year_of_production"]].groupby(["car_brand"]).mean()
df1 = df1.sort_values(by=["price"], ascending=False)
df1.head(100)
```

```python
px.scatter(df1, x="year_of_production", y="price", trendline="ols",
           title="Mean price by year of production on brand level")
```
